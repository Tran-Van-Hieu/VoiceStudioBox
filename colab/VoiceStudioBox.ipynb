# Voice Studio Box trên Google Colab

1. Menu **Thời gian chạy → Thay đổi loại thời gian chạy → T4 GPU → Lưu**.
2. Bấm nút ▶ ở ô bên dưới, cho phép truy cập Google Drive, chờ khoảng 15–20 phút lần đầu (tải sẵn OmniVoice và Qwen3-TTS 1.7B vào Drive), các lần sau nhanh hơn.
3. Khi hiện dòng `Open this link in your browser`, mở link `https://...trycloudflare.com` đó. App chạy ngay trong trình duyệt.

Giọng đã clone và lịch sử được sao lưu vào Google Drive (thư mục `VoiceStudioBox`) mỗi phút, lần mở sau vẫn còn. Link cũng được ghi vào `VoiceStudioBox/link.txt`. Model, thư viện và giao diện được giữ trong thư mục `VoiceStudioBox-cache` (khoảng 10 GB) để lần sau không tải lại. Khi có phiên bản model mới, log sẽ báo; vào **Cài đặt → Model** để chọn cập nhật hay giữ bản cũ.

Ai có link đều dùng được đầy đủ app và giọng của bạn (kể cả xóa), chỉ gửi cho người tin cậy. Link đổi mỗi lần chạy. Xong việc bấm ■ để dừng: app lưu lần cuối vào Drive rồi tắt.

```
# @title Chạy Voice Studio Box
luu_vao_google_drive = True  # @param {type:"boolean"}
# Bật: giọng đã clone và lịch sử được sao lưu vào Google Drive, lần sau vẫn còn.

import os, subprocess

if subprocess.run(['nvidia-smi'], capture_output=True).returncode != 0:
    raise SystemExit('Chưa bật GPU: Thời gian chạy → Thay đổi loại thời gian chạy → T4 GPU.')

persist = ''
if luu_vao_google_drive:
    from google.colab import drive
    try:
        drive.mount('/content/drive')
    except ValueError:
        raise SystemExit('Không kết nối được Google Drive. Bấm ▶ lại, chọn tài khoản và tick "Chọn tất cả" quyền, '
                         'hoặc bỏ chọn luu_vao_google_drive để chạy không lưu.')
    persist = ('--persist-dir /content/drive/MyDrive/VoiceStudioBox'
               ' --cache-dir /content/drive/MyDrive/VoiceStudioBox-cache')

if not os.path.exists('/content/VoiceStudioBox'):
    !git clone --depth 1 https://github.com/Tran-Van-Hieu/VoiceStudioBox.git /content/VoiceStudioBox
else:
    !git -C /content/VoiceStudioBox pull -q
%cd /content/VoiceStudioBox
!apt-get -qq install -y sox > /dev/null
!python scripts/remote_server.py --data-dir /content/vsb-data $persist
```
