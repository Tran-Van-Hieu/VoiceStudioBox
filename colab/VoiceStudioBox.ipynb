# Voice Studio Box trên Google Colab

2. Bấm nút ▶ ở ô bên dưới và chờ khoảng 5–10 phút lần đầu.
3. Khi hiện dòng `Open this link in your browser`, mở đường link `https://...trycloudflare.com` đó. App chạy ngay trong trình duyệt.
